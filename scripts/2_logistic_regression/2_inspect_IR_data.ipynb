```python
%load_ext autoreload
%autoreload 2

import sys
import os
from pathlib import Path
notebook_dir = Path().resolve()
sys.path.append(str(notebook_dir.parent))        # ../
sys.path.append(str(notebook_dir.parent.parent)) # ../../


import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../../data/insulin_resistance/IR_metrics.csv")

display(df.head())

#exclude DEC_42 from analysis due to less than 7 days of smartwatch data
df = df[df['study_id'] != 'DEC_42']

#*Histograms of residuals
plt.figure(figsize=(6, 4))

sns.histplot(df['matsuda_2h'], kde=True)

plt.xlabel("Matsuda_Index")
plt.title("Distribution of Matsuda_Index")
plt.tight_layout()
plt.show()

print(df["matsuda_2h"].describe())
print("IQR: ", 3.676031 - 1.903538)
```

```python
df["lower_IS"] = (
    df['matsuda_2h'] <= 4.0
).astype(int)

n_IR = df['lower_IS'].sum()
n_IS = (df['lower_IS'] == 0).sum()
n_patients= df['study_id'].nunique()

print(n_patients, "patients in total")

print("Number of patients with Matsuda Index >4:", n_IS, f"({n_IS/n_patients*100:.2f}%)")
print("Number of patients with Matsuda Index <=4:", n_IR, f"({n_IR/n_patients*100:.2f}%)")
```

```python
plt.figure(figsize=(6, 4))

sns.displot(df['matsuda_2h'], kde=True, )
#add cutoff line at 4
plt.axvline(x=4, color='red', linestyle='--', label='Cutoff at 4')
plt.xlabel("Matsuda_Index")
plt.title("Distribution of Matsuda_Index with Cutoff")
plt.legend()

print("Number of patients with Matsuda Index = 4:", (df['matsuda_2h'] == 4.0).sum())
```

```python
for distance in [0.25, 0.5, 1.0]:
    n_close = df["matsuda_2h"].between(
        4 - distance,
        4 + distance
    ).sum()

    print(f"Within ±{distance} of cutoff: {n_close}")
```
